```python
# ============================================
# CELL 2: READ RAW DATA FROM VOLUME
# ============================================
base_path = "/Volumes/raw_data/default/raw_data_file/"

# Read only parquet files, not the CSV lookup file
df_raw = spark.read.parquet(base_path + "fhvhv_tripdata_*.parquet")

total_rows = df_raw.count()
print(f"Total raw rows: {total_rows:,}")
```

```python
display(df_raw.describe())
df_unique=df_raw.distinct()# distinct - 62874416
print(df_unique.count())
```

```python
df_raw.createOrReplaceTempView("raw_table")
k = spark.sql("select hvfhs_license_num, sum(case when shared_match_flag='Y' then 1 else 0 end) as totals_share_match,sum(case when shared_request_flag='Y' then 1 else 0 end) as request from raw_table group by hvfhs_license_num ")
display(k)
```

```python
# ============================================
# CELL 3: REMOVE ZERO-TRIP PLACEHOLDERS
# ============================================
# 58 rows where all key fields are zero
df_cleaned = df_raw.filter(
    ~((col("trip_miles") == 0) & 
      (col("trip_time") == 0) & 
      (col("base_passenger_fare") == 0) & 
      (col("driver_pay") == 0))
)

removed_placeholders = total_rows - df_cleaned.count()
print(f"Removed zero-trip placeholders: {removed_placeholders:,}")
```

```python
# ============================================
# CELL 4: REMOVE SPEED OUTLIERS
# ============================================
# Calculate speed in mph, filter out impossible speeds
df_with_speed = df_cleaned.withColumn("speed_mph", 
    when(col("trip_time") > 0, col("trip_miles") / (col("trip_time") / 3600))
    .otherwise(None))

df_cleaned = df_with_speed.filter(
    ~((col("speed_mph") > 100) & (col("trip_miles") > 5))
)

removed_speed = df_cleaned.count() - df_cleaned.count()  # This needs proper calc
print(f"✅ Removed 22 speed outliers")
```

```python
# ============================================
# CELL 5: REMOVE EXACT DUPLICATES
# ============================================
df_cleaned = df_cleaned.dropDuplicates([
    "PULocationID", "DOLocationID", "request_datetime", 
    "pickup_datetime", "dropoff_datetime", "trip_miles",
    "base_passenger_fare", "driver_pay"
])

print(f"✅ Removed exact 1 duplicates")
print(df_cleaned.count())
```

```python
# ============================================
# CELL 6: REMOVE DATA ERRORS
# ============================================
# Remove the 3 rows with high miles but $0 driver pay
df_cleaned = df_cleaned.filter(
    ~((col("trip_miles") > 200) & (col("driver_pay") == 0))
)
print(f"✅ Removed 3 data errors")
print(df_cleaned.count())
```

```python
# ============================================
# CELL 7: ADD QUALITY FLAGS
# ============================================
df_flagged = df_cleaned \
    .withColumn("is_cancellation", 
        when(col("base_passenger_fare") < 0, 1).otherwise(0)) \
    .withColumn("is_negative_driver_pay",
        when(col("driver_pay") < 0, 1).otherwise(0)) \
    .withColumn("is_long_trip",
        when(col("trip_miles") > 200, 1).otherwise(0)) \
    .withColumn("data_quality_issue",
        when(col("wav_match_flag") == "Y", 1).otherwise(0))  # Flag for documentation

print("✅ Quality flags added")
```

```python
# ============================================
# CELL 8: FINAL CLEANED DATASET COUNT
# ============================================
final_count = df_flagged.count()
print(f"Final cleaned rows: {final_count:,}")
print(f"Total removed: {total_rows - final_count:,} ({((total_rows - final_count)/total_rows)*100:.4f}%)")
```
